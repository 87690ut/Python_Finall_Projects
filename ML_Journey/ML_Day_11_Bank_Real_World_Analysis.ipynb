```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification

# 1. generating imbalance data (95% good, 5% Fraud)
X_data, y_data = make_classification(n_samples=1000, n_features=4, 
                                     weights=[0.95, 0.05], random_state=42)

df = pd.DataFrame(X_data, columns=['Income', 'Loan_Amount', 'Credit_Score', 'Months_Employed'])
df['Target'] = y_data # 0 = Loan Paid, 1 = Fraud

# 2.(Missing values/NaN) insertion
df.loc[10:25, 'Income'] = np.nan
df.loc[70:85, 'Credit_Score'] = np.nan

```

### Data checking.

```python
# 3. Data checking
print("--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Target Balance (Imbalance Check) ---")
print(df['Target'].value_counts())   # it will count total unique values.
```

```python
df['Income'] = df['Income'].fillna(df['Income'].median())   # removing null values from Income column
df['Income'].isnull().sum()
```

```python
df['Credit_Score'] = df['Credit_Score'].fillna(df['Credit_Score'].median())
df['Credit_Score'].isnull().sum()
```
