```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
data = {
    'Age': [20,25,40,50,22,45],
    'Discount' : [5,10,20,30,30,5],
    'Sale': [0,0,1,1,1,0]
}
df = pd.DataFrame(data)
df.head()
```

```python
df['Age'] = df['Age'].astype(float)
```

```python
df.info()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
import matplotlib.pyplot as ply
from sklearn.tree import plot_tree
```

```python
X = df[['Age','Discount']]
y = df['Sale']

model = DecisionTreeClassifier()
model.fit(X,y)
print("Data reading completed.")
```

```python
plt.figure(figsize=(10,6))
plot_tree(model,feature_names=['Age','Discount'],class_names=['Not purchased (0)','Purchased (1)'], filled = True, rounded = True)
plt.show()
```
